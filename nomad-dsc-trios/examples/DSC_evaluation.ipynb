# DSC evaluation (tangent method) with the mouse

1. Run the cell below (Shift+Enter).
2. Choose the **folder** and the **curve** (◀ ▶ step through the folder).
3. **Drag across a peak** in the cooling (top) or heating (bottom) plot, from the baseline point on one side to the baseline point on the other: the window is set and the tangents are recomputed at once. The other ramp keeps the automatic windows.
4. **Save to NOMAD** writes mode, cycle and windows into the DSC entry; reload the entry page in NOMAD.
5. **Results of this folder** shows the values stored in NOMAD and saves them as `0000_dsc_results.csv` in the folder.

Values typed in by hand in NOMAD are never overwritten.
